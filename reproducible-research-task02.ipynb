```python
import subprocess
import sys

# Run pip freeze using Python's internal subprocess
result = subprocess.run([sys.executable, '-m', 'pip', 'freeze'], capture_output=True, text=True)

# Write the output directly to the file
with open('requirements.txt', 'w') as file:
    file.write(result.stdout)

print("requirements.txt successfully created!")
```

```python
import pandas as pd
import numpy as np

# 1. Define the data pipeline based on your preregistration rules
def clean_retail_data(df):
    # Rule 1: Drop missing (null) values in target columns
    df_clean = df.dropna(subset=['Product Category', 'Total Amount'])
    # Rule 2: Exclude anomalous transactions where Total Amount is 0 or negative
    df_clean = df_clean[df_clean['Total Amount'] > 0]
    return df_clean

# 2. Create synthetic dummy data with deliberate errors
synthetic_data = pd.DataFrame({
    'Transaction ID': [1, 2, 3, 4, 5],
    'Product Category': ['Beauty', 'Clothing', np.nan, 'Electronics', 'Clothing'],
    'Total Amount': [150.0, -50.0, 100.0, 200.0, 0.0] 
})

print("--- Original Synthetic Data ---")
print(synthetic_data)

# 3. Run the data-blind pipeline check
processed_data = clean_retail_data(synthetic_data)

print("\n--- Processed Data (Should only keep Transactions 1 and 4) ---")
print(processed_data)
```
